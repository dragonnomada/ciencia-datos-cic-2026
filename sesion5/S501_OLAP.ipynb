# OLAP - Cubo Discreto

```python
import numpy
import pandas

from matplotlib import pyplot
import seaborn
```

```python
titanic = pandas.read_excel("Titanic.xlsx", sheet_name="Titanic")

titanic
```

```python
def survived_clasess(survived):
  if survived == 0:
    return "No"
  else:
    return "Yes"

d1 = titanic["Survived"].apply(survived_clasess)

d1
```

```python
def pclass_clasess(pclass):
  if pclass == 1:
    return "1st"
  elif pclass == 2:
    return "2nd"
  elif pclass == 3:
    return "3rd"
  else:
    return "unknown"

d2 = titanic["Pclass"].apply(pclass_clasess)

d2
```

```python
def age_classes(age):
  if numpy.isnan(age):
    return "unknown"

  # Limpieza de datos
  if age < 1:
    age = 100 * age # Limpiamos las edades menores a 1

  if age <= 10:
    return "children"
  elif age <= 19:
    return "young"
  elif age <= 25:
    return "adult1"
  elif age <= 35:
    return "adult2"
  elif age <= 45:
    return "adult3"
  elif age <= 55:
    return "adult4"
  else:
    return "adulto5"

d3 = titanic["Age"].apply(age_classes)

d3
```

```python
d4 = titanic["Sex"]
d4
```

```python
X = pandas.DataFrame([d1, d2, d3, d4]).T

X
```

```python
classes = ["Survived"]

S = X[classes].groupby(classes).size()
P = 100 * S / S.sum()

pandas.concat([S, P], axis=1, keys=["Support", "Probability"]).round(1)
```

```python
classes = ["Pclass"]

S = X[classes].groupby(classes).size()
P = 100 * S / S.sum()

pandas.concat([S, P], axis=1, keys=["Support", "Probability"]).round(1)
```

```python
classes = ["Sex"]

S = X[classes].groupby(classes).size()
P = 100 * S / S.sum()

pandas.concat([S, P], axis=1, keys=["Support", "Probability"]).round(1)
```

```python
classes = ["Age"]

S = X[classes].groupby(classes).size()
P = 100 * S / S.sum()

pandas.concat([S, P], axis=1, keys=["Support", "Probability"]).round(1)
```

```python
classes = ["Survived", "Pclass"]

S = X[classes].groupby(classes).size()
P = 100 * S / S.sum()

pandas.concat([S, P], axis=1, keys=["Support", "Probability"]).round(1)
```

```python
classes = ["Survived", "Pclass"]

S = X[classes].groupby(classes).size()
P = 100 * S / S.sum()

T = pandas.concat([S, P], axis=1, keys=["Support", "Probability"]).round(1).reset_index()

T
```
