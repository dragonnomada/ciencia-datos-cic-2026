```python
# Circulo: con centro (7, 8) con radio 3
# Punto: en (5, 5)

ox = 7
oy = 8
r = 3

px = 5
py = 5

dx = px - ox
dy = py - oy

# d^2 = dx^2 + dy^2
# d = (dx^2 + dy^2)^(1 / 2)
d = (dx ** 2 + dy ** 2) ** 0.5 # C^2 = A^2 + B^2

esta_dentro = d <= r # True - Dentro / False - Fuera

esta_dentro
```

```python
# 1.  ---------**---------   - x 9 | * x 1 | * x 1 | - x 9
# 2.  --------****--------   - x 8 | * x 2 | * x 2 | - x 8
# 3.  -------******-------   - x 7 | * x 3 | * x 3 | - x 7
# 4.  ------********------   ...
# 5.  -----**********-----   ...
# 6.  ----************----   ...
# 7.  ---**************---   - x 3 | * x 7 | * x 7 | - x 3
# 8.  --****************--   ...
# 9.  -******************-   ...
# 10. ********************   - x 0 | * x 10 | * x 10 | - x 0

# i.  ????????????????????   - x (10 - i) | * x i | * x i | - x (10 - i)

# Secuenciador: PARA <elemento> EN <secuencia/rango/lista/colección>,
#                 ENTONCES -> BLOUQE

for i in range(1, 11): # i = 1,2,3,...,10 | range(a, b) -> a...(b-1)
  # i = 1 ->
  print("-" * (10 - i) + "*" * i + "*" * i + "-" * (10 - i))
```

```python
# Función: Un bloque parametrizado reutilizable
# (<parámetros>) -> BLOQUE
# Sirven para abstraer código y formalizar cálculos y tareas rutinarias
# donde solo cambien algunos parámetros

# Solamente define la función
# Es un bloque de 4 líneas de códigos
# Y requiere 4 parámetros
def distanciaEntreDosPuntos(x1, y1, x2, y2):
  dx = x2 - x1
  dy = y2 - y1

  d = (dx ** 2 + dy ** 2) ** 0.5

  return d
```

```python
distanciaEntreDosPuntos(1, 2, 3, 4) # (1, 2) -> (3, 4)
```

```python
distanciaEntreDosPuntos(0, 0, 3, 4) # (0, 0) -> (3, 4)
```

```python
# Colecciones: Listas, Tuplas, Diccionarios

# Listas: elementos indexados extendibles
# Tuplas: elementos indexados fijos
# Diccionarios: elementos asociados a claves extendibles

# Listas: [<elemento1>, <elemento2>, ..., <elementoN>]
# Tuplas: (<elemento1>, <elemento2>, ..., <elementoK>)
# Diccionarios:
# {
#    "<clave1>": <valor1>,
#    "<clave2>": <valor2>,
#    ...,
#    "<claveP>": <valorP>
# }

# Lista de 3 frutas (3-textos)
frutas = ["manzana", "pera", "kiwi"]
print(frutas)
# Agregar una fruta al final
frutas.append("mango")
print(frutas)
# Agregar una fruta en un índice específico (del 0 hacia adelante)
frutas.insert(2, "piña") # -> Desplaza los demás elementos a la derecha
print(frutas)
```

```python
# Tuplas: Acoplar valores para transportarlos y desacoplarlos en variables nuevas

circulo = (7, 8, 3) # 3-valores | 3-tupla

punto = (5, 5) # 2-valores | 2-tupla
```

```python
ox, oy, radio = circulo # Desacoplar en variables (3-tupla -> 3-variables)

px, py = punto # 2-tupla -> 2-varibles
```

```python
def estaPuntoDentroCirculo2(punto, circulo):
  px, py = punto
  ox, oy, r = circulo

  dx = px - ox
  dy = py - oy

  d = (dx ** 2 + dy ** 2) ** 0.5

  return d <= r
```

```python
estaPuntoDentroCirculo2((5, 5), (7, 8, 3))
```
